Message to Encode -> Frequency -> Huffman -> Varcode -> Encode

```python
class Node:
    def __init__(self, freq, symbol=None):
        self.__freq = 0
        self.__symbol = symbol
        self.__left = None
        self.__right = None

    def get_left(self):
        return self.__left

    def get_right(self):
        return self.__right

    def set_left(self, node):
        self.__left = node

    def set_right(self, node):
        self.__right = node

    def get_freq(self):
        return self.__freq

    def get_symbol(self):
        return self.__symbol

    def __lt__(self, other):
        return self.__freq < other.get_freq()
```

```python
class Huffman:

    #L/R values
    LEFT = 1
    RIGHT = 0

    def __init__(self):
        # frequency table
        self.__freq = dict()
        # forest of nodes
        self.__forest = []
        # root of huffman tree
        self.__huffman_root = None

    def compute_frequencies(self, message):
        for char in message:
            if char in self.__freq:
                self.__freq[char] += 1
            else:
                self.__freq[char] = 1

    def initialize_forest(self):
        for symbol in self.__freq:
            # create a node for this symbol and its frequency
            node = Node(self.__freq[symbol], symbol)
            # add the node to the forest
            self.__forest.append(node)

    def build_huffman_tree(self):
        while len(self.__forest) > 1:
            # find the two smallest trees in the forest
            small1 = self.smallest_node()
            small2 = self.smallest_node()
            # and combine them with None symbol and combination of freq
            combined_freq = small1.get_freq() + small2.get_freq()
            # create a new node without symbol
            new_node = Node(combined_freq)
            # assign children to it
            new_node.set_left(small1)
            new_node.set_right(small2)
            # place this node to the forest
            self.__forest.append(new_node)
            # what we have left is the root of the huffman tree
        self.__huffman_root = self.__forest[0]

    def smallest_node(self):
        smallest_index = 0
        for i in range(len(self.__forest)):
            if self.__forest[i] < self.__forest[smallest_index]:
                smallest_index = i
        smallest = self.__forest.remove(smallest_index)
        return smallest


    def produce_varcodes(self, ):
        codes = {}
        stack=[(self.__huffman_root,"")]
        while stack:
            node, path = stack.pop()
            if node.get_symbol() != None:
                codes[node.get_symbol()] = path
            else:
                if node.has_left():
                    stack.append(node.get_left(), path + LEFT)
                if node.has_right():
                    stack.append(node.get_right(), path + RIGHT)
```
